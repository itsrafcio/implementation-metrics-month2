```python
import pandas as pd
```

```python
projects = pd.read_csv("../data/projects.csv")
customers = pd.read_csv("../data/customers.csv")
milestones = pd.read_csv("../data/milestones.csv")
costs = pd.read_csv("../data/costs.csv")
```

```python
projects_with_customers = projects.merge(
    customers,
    left_on="customer_id",
    right_on="id",
    how="left",
    suffixes=("_projects", "_customers"),
    validate="many_to_one"
)
projects_with_customers.head()
```
